```python
# 이항 분류
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
from tensorflow.keras.optimizers import Adam
```

```python
x_data=np.array([[1,2],[2,3],[3,4],[4,3],[3,2],[2,1]],dtype=np.float32)
y_data=np.array([[0],[0],[0],[1],[1],[1]],dtype=np.float32)

# 모델 정의 방법1(Sequential api)
# model=Sequential([
#     Input(np.shape=(2,)),
#     Dense(units=1,activation='sigmoid'),
# ])
model = Sequential()                     # 층을 순서대로 쌓는 컨테이너(Sequential API)
model.add(Input(shape=(2,)))             # 입력 텐서: 각 샘플이 특징 2개를 가짐 → (batch, 2)

model.add(Dense(units=1, activation='sigmoid')) # - Dense의 파라미터 수: (입력차원 2 × 유닛 1) + 바이어스 1 = 총 3개 / sigmoid 출력은 0~1 확률로 해석 (양성 클래스일 확률)

# 컴파일(손실/옵티마이저/지표 설정)
model.compile(
    loss='binary_crossentropy',           # - loss='binary_crossentropy': y가 0/1(스칼라)인 이진 분류 손실 / 출력층이 sigmoid일 때 사용 (from_logits=False 기본 가정)
    optimizer=Adam(learning_rate=0.01),   # - optimizer=Adam(learning_rate=0.01): 적응적 학습률을 쓰는 Adam 옵티마이저
    metrics=['accuracy']                  # - metrics=['accuracy']: 평가 시 정확도를 함께 출력
)

model.summary()

history = model.fit(
    x_data, y_data,
    epochs=50,      # - epochs=50    : 전체 학습 데이터를 50번 반복 학습
    batch_size=1,   # - batch_size=1 : 샘플 1개 단위로 가중치 업데이트(SGD와 유사한 효과)
    verbose=2       # - verbose=2    : 에폭마다 한 줄 로그 형식
)
m_eval=model.evaluate(x_data,y_data,verbose=0)
print(f'평가 결과 : 손실={m_eval[0]:.4f}, 정확도={m_eval[1]:.4f}')  # 평가 결과 : 손실=0.4659, 정확도=1.0000

new_data=np.array([[1,2.5],[10.5,7.1]],dtype=np.float32)
pred=model.predict(new_data,verbose=0)
print('예측확률: ',pred.ravel())    # 예측확률:  [0.7294258  0.14271492]
print('예측결과: ',[1 if i>=0.5 else 0 for i in pred])  # 예측결과:  [1, 0]
print('예측결과: ',(pred>=0.5).astype(int).ravel())     # 예측결과:  [1 0]
print('예측결과: ',np.where(pred>=0.5,1,0).ravel())     # 예측결과:  [1 0]
```

```python
# 모델 정의 방법2 : Funcional api
from tensorflow.keras.models import Model

input_layer=Input(shape=(2,))
output_layer=Dense(1,activation='sigmoid')(input_layer)

model2 = Model(inputs=input_layer, outputs=output_layer)

# 컴파일(손실/옵티마이저/지표 설정)
model2.compile(
    loss='binary_crossentropy',           # - loss='binary_crossentropy': y가 0/1(스칼라)인 이진 분류 손실 / 출력층이 sigmoid일 때 사용 (from_logits=False 기본 가정)
    optimizer=Adam(learning_rate=0.01),   # - optimizer=Adam(learning_rate=0.01): 적응적 학습률을 쓰는 Adam 옵티마이저
    metrics=['accuracy']                  # - metrics=['accuracy']: 평가 시 정확도를 함께 출력
)

model2.summary()

model2.fit(
    x_data, y_data,
    epochs=50,      # - epochs=50    : 전체 학습 데이터를 50번 반복 학습
    batch_size=1,   # - batch_size=1 : 샘플 1개 단위로 가중치 업데이트(SGD와 유사한 효과)
    verbose=2       # - verbose=2    : 에폭마다 한 줄 로그 형식
)
m_eval=model2.evaluate(x_data,y_data,verbose=0)
print(f'평가 결과 : 손실={m_eval[0]:.4f}, 정확도={m_eval[1]:.4f}')  # 평가 결과 : 손실=0.4659, 정확도=1.0000

new_data=np.array([[1,2.5],[10.5,7.1]],dtype=np.float32)
pred=model2.predict(new_data,verbose=0)
print('예측확률: ',pred.ravel())
print('예측결과: ',[1 if i>=0.5 else 0 for i in pred])
```

```python
# 모델 정의 방법3 : Subclassing api
class MyBinaryClass(Model):
    def __init__(self):
      super().__init__(name='MyBinaryClass')
      self.dense=Dense(1,activation='sigmoid',name='dense_sigmoid')

    def build(self,input_shape):{
        # 첫번째 feed forward에서 가중치(w,b)를 만든다.
        super().build(input_shape)
    }

    # fit(), evaluate(), predict() 하면 자동 호출
    def call(self, inputs, training=False):
      print('>>> call 실행됨, training=',training)
      return self.dense(inputs)

model3=MyBinaryClass()
model3.build(input_shape=(None,2))
model3.summary()

# 컴파일(손실/옵티마이저/지표 설정)
model3.compile(
    loss='binary_crossentropy',           # - loss='binary_crossentropy': y가 0/1(스칼라)인 이진 분류 손실 / 출력층이 sigmoid일 때 사용 (from_logits=False 기본 가정)
    optimizer=Adam(learning_rate=0.01),   # - optimizer=Adam(learning_rate=0.01): 적응적 학습률을 쓰는 Adam 옵티마이저
    metrics=['accuracy']                  # - metrics=['accuracy']: 평가 시 정확도를 함께 출력
)

model3.fit(
    x_data, y_data,
    epochs=50,      # - epochs=50    : 전체 학습 데이터를 50번 반복 학습
    batch_size=1,   # - batch_size=1 : 샘플 1개 단위로 가중치 업데이트(SGD와 유사한 효과)
    verbose=2       # - verbose=2    : 에폭마다 한 줄 로그 형식
)
m_eval=model3.evaluate(x_data,y_data,verbose=0)
print(f'평가 결과 : 손실={m_eval[0]:.4f}, 정확도={m_eval[1]:.4f}')  # 평가 결과 : 손실=0.4659, 정확도=1.0000

new_data=np.array([[1,2.5],[10.5,7.1]],dtype=np.float32)
pred=model3.predict(new_data,verbose=0)
print('예측확률: ',pred.ravel())
print('예측결과: ',[1 if i>=0.5 else 0 for i in pred])
```
